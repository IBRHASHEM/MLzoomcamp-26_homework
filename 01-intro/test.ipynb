```python
import pandas as pd
import numpy as np

data = {
    'car_name': ['Toyota', 'Honda', 'Ford', 'BMW', 'Nissan'],
    'origin': ['Asia', 'Asia', 'USA', 'Europe', 'Asia'],
    'horsepower': [130, np.nan, 250, np.nan, 140],
    'mpg': [32, 35, 18, 22, 30]
}

df = pd.DataFrame(data)
df
```

```python
print("Missing:", df['horsepower'].isnull().sum())
```

```python
mean_val = df['horsepower'].mean()
df['horsepower'] = df['horsepower'].fillna(mean_val)
max_asia_mpg = df[df['origin'] == 'Asia']['mpg'].max()
print("Max Asia MPG:", max_asia_mpg)
```
