```python
median_before = df['horsepower'].median()
print(f"Median before: {median_before}")
```

```python
mode_val = df['horsepower'].mode()[0]
print(f"Most frequent value (Mode): {mode_val}")
```

```python
df_filled = df['horsepower'].fillna(mode_val)
df_filled
```

```python
median_after = df_filled.median()
print(f"Median after: {median_after}")
```

```python
print(f"Has it changed? {median_before != median_after}")
```

```python
import numpy as np
```

```python
df_asia = df[df['origin'] == 'Asia'][['vehicle_weight', 'model_year']]
```

```python
X = df_asia.iloc[:7].values
```

```python
XTX = X.T.dot(X)
```

```python
XTX_inv = np.linalg.inv(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
```

```python
w = XTX_inv.dot(X.T).dot(y)
```

```python
w_sum = w.sum()
print("Result:", w_sum)
print("Rounded result:", round(w_sum, 4))
```
